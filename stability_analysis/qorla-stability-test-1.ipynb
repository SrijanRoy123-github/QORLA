```python
import torch, pandas as pd, numpy as np, matplotlib.pyplot as plt
from IPython.display import display, Markdown

torch.manual_seed(7)
N, lam, rho0, eps = 256, 4.0, 1e-3, 1e-8
TAUS = [.01,.02,.03,.05,.08,.10]

# ----------------------------- QORLA ----------------------------------------
def qorla(a,b,v,mode="fixed",tau=.03,beta=.1,gamma=1e-3):
    z=b[None,:]-a[:,None]; W=torch.exp(-lam*z.abs())
    S0=W.sum(1); S1=(W*z).sum(1); S2=(W*z*z).sum(1)
    mu=S1/(S0+eps); cz=z-mu[:,None]; Q=(W*cz.square()).sum(1)
    T0=W@v; C=(W*cz)@v
    rho=torch.full_like(S0,rho0)
    var=(S2/(S0+eps)-(S1/(S0+eps)).square()).clamp_min(0)

    if mode=="variance": rho=rho0+beta*var
    if mode=="mass": rho=rho0+beta*var+gamma/(S0+eps)
    if mode=="hard":
        req=S1.square()/(S0*(1-tau)+eps)-S2
        rho=torch.maximum(rho,req)

    base=T0/(S0+eps)[:,None]
    corr=-mu[:,None]*C/(Q+rho+eps)[:,None]
    k=(Q+rho)/(S2+rho+eps)

    if mode=="backward":
        s=(k.detach()/tau).clamp(max=1)
        corr=corr.detach()+s[:,None]*(corr-corr.detach())
    else: s=torch.ones_like(k)

    out=base+corr
    D=S0*(S2+rho)-S1.square()+eps
    al=W*(S2[:,None]+rho[:,None]-S1[:,None]*z)/D[:,None]
    return out,al,k,rho,s,D

# ------------------------- Stress geometries -------------------------------
def geom(name,n=N,dv=8,seed=1):
    torch.manual_seed(seed)
    if name=="normal": a,b=torch.randn(n),torch.randn(n)
    elif name=="clustered": a,b=.01*torch.randn(n),.001*torch.randn(n)
    elif name=="boundary":
        b=torch.linspace(0,1,n)
        a=torch.cat([torch.linspace(-.2,.15,n//2),torch.linspace(.85,1.2,n-n//2)])
    elif name=="low_mass": b=.1*torch.randn(n); a=1.5+.1*torch.randn(n)
    else:
        b=torch.linspace(-1,1,n)+1e-6*torch.randn(n); a=b+1e-6*torch.randn(n)
    return a,b,torch.randn(n,dv)

cases=["normal","clustered","boundary","low_mass","near_ties"]
methods=[("Fixed","fixed"),("Variance ridge","variance"),("Mass ridge","mass"),
         ("Hard floor","hard"),("Backward damping","backward")]

stress=[]
for ci,c in enumerate(cases):
    A,B,V=geom(c,seed=100+ci)
    for label,mode in methods:
        a=A.clone().requires_grad_(); b=B.clone().requires_grad_(); v=V.clone().requires_grad_()
        o,al,k,r,s,D=qorla(a,b,v,mode,.03); L=o.square().mean(); L.backward()
        stress.append([c,label,L.item(),D.min().item(),k.min().item(),al.abs().max().item(),
                       (al.sum(1)-1).abs().max().item(),a.grad.norm().item(),
                       b.grad.norm().item(),v.grad.norm().item()])
stress=pd.DataFrame(stress,columns=[
    "case","method","loss","min_D","min_kappa","max_alpha",
    "alpha_sum_error","grad_a","grad_b","grad_v"
])

# ------------------------ Affine preservation -------------------------------
aff=[]
for label,mode in methods:
    b=torch.linspace(-1,1,N); a=torch.linspace(-1.2,1.2,N)
    v=(.7+1.3*b)[:,None]; y=(.7+1.3*a)[:,None]
    o,al,k,*_=qorla(a,b,v,mode,.03)
    aff.append([label,(o-y).abs().mean().item(),(o-y).abs().max().item(),k.min().item()])
aff=pd.DataFrame(aff,columns=["method","MAE","max_error","min_kappa"])

# ------------------- Why τg = 0.03: fidelity sweep --------------------------
fa=lambda x:.7+1.3*x
fq=lambda x:.7+1.3*x+2*x.square()
torch.manual_seed(7)
G={
 "cluster_affine":(.01*torch.randn(N),.001*torch.randn(N),fa),
 "cluster_quad":(.01*torch.randn(N),.001*torch.randn(N),fq),
 "lowmass_affine":(1.5+.1*torch.randn(N),.1*torch.randn(N),fa),
 "lowmass_quad":(1.5+.1*torch.randn(N),.1*torch.randn(N),fq)
}

def grads(A,B,V,Y,t=None):
    a=A.clone().requires_grad_(); b=B.clone().requires_grad_(); v=V.clone().requires_grad_()
    o,*_=qorla(a,b,v,"fixed" if t is None else "backward",.03 if t is None else t)
    L=((o-Y)**2).mean(); L.backward()
    return a.grad.detach(),b.grad.detach(),v.grad.detach()

def cosine(x,y): return (x.flatten()@y.flatten()/(x.norm()*y.norm()+1e-30)).item()

F=[]
for name,(A,B,f) in G.items():
    V=f(B)[:,None]; Y=f(A)[:,None]; ga0,gb0,gv0=grads(A,B,V,Y)
    for t in TAUS:
        ga,gb,gv=grads(A,B,V,Y,t)
        F.append([name,t,cosine(ga0,ga),cosine(gb0,gb),cosine(gv0,gv),
                  (ga.norm()/ga0.norm()).item(),(gb.norm()/gb0.norm()).item(),
                  (gv.norm()/gv0.norm()).item()])

fid=pd.DataFrame(F,columns=[
    "test","tau","cos_a","cos_b","cos_v","grad_a_ratio","grad_b_ratio","grad_v_ratio"
])

sel=fid.groupby("tau").agg(
    mean_cos_a=("cos_a","mean"), min_cos_a=("cos_a","min"),
    mean_cos_b=("cos_b","mean"), min_cos_b=("cos_b","min"),
    mean_cos_v=("cos_v","mean"), min_cos_v=("cos_v","min"),
    mean_grad_b_ratio=("grad_b_ratio","mean")
).reset_index()

# Fidelity requirements: retain useful gradient direction, then maximize damping
sel["fidelity_pass"]=(sel.min_cos_a>=.90)&(sel.min_cos_b>=.95)&(sel.min_cos_v>=.95)
chosen=sel[sel.fidelity_pass].sort_values("mean_grad_b_ratio").iloc[0]
TAU_SELECTED=float(chosen.tau)

# ------------------------ Multi-seed confirmation ---------------------------
def train_case(kind,t,seed,steps=100,n=96):
    torch.manual_seed(seed)
    if kind=="normal": A,B=torch.randn(n),torch.randn(n)
    elif kind=="cluster": A,B=.01*torch.randn(n),.001*torch.randn(n)
    else: A,B=1.5+.1*torch.randn(n),.1*torch.randn(n)
    Y=(.7+1.3*A+2*A.square())[:,None]
    a=A.clone().requires_grad_(); b=B.clone().requires_grad_(); v=torch.randn(n,1).requires_grad_()
    opt=torch.optim.Adam([a,b,v],lr=3e-3); losses=[]; gb=[]; bad=False

    for _ in range(steps):
        opt.zero_grad()
        o,*_=qorla(a,b,v,"fixed" if t is None else "backward",TAU_SELECTED if t is None else t)
        L=((o-Y)**2).mean()
        if not torch.isfinite(L): bad=True; break
        L.backward(); losses.append(L.item()); gb.append(b.grad.norm().item()); opt.step()

    return [kind,"fixed" if t is None else t,seed,losses[-1],max(gb),np.mean(gb),bad]

rob=pd.DataFrame(
    [train_case(c,t,s) for c in ["normal","cluster","lowmass"]
     for t in [None,.03,.05] for s in [1,3,7,11,19]],
    columns=["case","tau","seed","final_loss","max_grad_b","mean_grad_b","failed"]
)

rob_sum=rob.groupby(["case","tau"],sort=False).agg(
    mean_final_loss=("final_loss","mean"),
    mean_max_grad_b=("max_grad_b","mean"),
    worst_grad_b=("max_grad_b","max"),
    mean_grad_b=("mean_grad_b","mean"),
    failures=("failed","sum")
).reset_index()

# ----------------------------- Report ---------------------------------------
display(Markdown(r"""
## QORLA Stability Test 1

\[
o_i=B_i+R_i,\qquad
B_i=\frac{T_{0,i}}{S_{0,i}+\epsilon},
\qquad
R_i=-\mu_i\frac{C_i}{Q_i+\rho+\epsilon}
\]

\[
\kappa_i=\frac{Q_i+\rho}{S_{2,i}+\rho+\epsilon}
\]

Condition-aware backward damping:

\[
s_i=\min\left(1,\frac{\operatorname{stopgrad}(\kappa_i)}{\tau_g}\right)
\]

\[
\widetilde R_i=\operatorname{sg}(R_i)+s_i[R_i-\operatorname{sg}(R_i)]
\]

The forward operator is unchanged. Only the unstable local-linear backward
contribution is damped.
"""))

print("STRESS METRICS")
display(stress.round(6))

print("\nAFFINE REPRODUCTION")
display(aff.round(8))

print("\nTHRESHOLD SELECTION")
display(sel.round(6))

display(Markdown(f"""
### How the threshold was selected

Candidate thresholds:

\[
\\tau_g\in\{{0.01,0.02,0.03,0.05,0.08,0.10\}}
\]

A threshold was considered admissible only if the worst gradient-direction
similarities satisfied

\[
\min\cos(g_a,\tilde g_a)\ge0.90,\qquad
\min\cos(g_b,\tilde g_b)\ge0.95,\qquad
\min\cos(g_v,\tilde g_v)\ge0.95.
\]

Among admissible thresholds, the value producing the **smallest mean
key-gradient norm ratio** was selected.

\[
\boxed{{\tau_g={TAU_SELECTED:.2f}}}
\]

For the selected value:

- mean key-gradient cosine = **{chosen.mean_cos_b:.4f}**
- worst key-gradient cosine = **{chosen.min_cos_b:.4f}**
- worst query-gradient cosine = **{chosen.min_cos_a:.4f}**
- worst value-gradient cosine = **{chosen.min_cos_v:.4f}**
- mean key-gradient norm ratio = **{chosen.mean_grad_b_ratio:.3f}**

Thus the selected threshold provides the strongest tested damping while
remaining inside the predefined gradient-fidelity region.
"""))

print("MULTI-SEED CONFIRMATION")
display(rob_sum.round(6))

# ------------------------------- Plots --------------------------------------

plt.figure(figsize=(7,4))
for m in stress.method.unique():
    d=stress[stress.method==m]
    plt.plot(d.case,d.grad_b,marker="o",label=m)
plt.ylabel(r"$||\nabla_b||$"); plt.title("Gradient stability across stress geometries")
plt.xticks(rotation=25); plt.legend(); plt.grid(alpha=.25); plt.tight_layout(); plt.show()

plt.figure(figsize=(7,4))
plt.plot(sel.tau,sel.mean_grad_b_ratio,marker="o",label="Mean gradient norm ratio")
plt.axvline(TAU_SELECTED,linestyle="--",label=f"Selected τ={TAU_SELECTED:.2f}")
plt.xlabel(r"$\tau_g$"); plt.ylabel("Gradient norm / baseline")
plt.title("Gradient damping versus threshold")
plt.legend(); plt.grid(alpha=.25); plt.tight_layout(); plt.show()

plt.figure(figsize=(7,4))
plt.plot(sel.tau,sel.min_cos_a,marker="o",label="min cosine: query")
plt.plot(sel.tau,sel.min_cos_b,marker="o",label="min cosine: key")
plt.plot(sel.tau,sel.min_cos_v,marker="o",label="min cosine: value")
plt.axhline(.90,linestyle="--"); plt.axhline(.95,linestyle=":")
plt.axvline(TAU_SELECTED,linestyle="--")
plt.xlabel(r"$\tau_g$"); plt.ylabel("Worst cosine similarity")
plt.title("Gradient-direction fidelity versus damping")
plt.legend(); plt.grid(alpha=.25); plt.tight_layout(); plt.show()

r=rob.groupby("tau",sort=False).agg(mean_max=("max_grad_b","mean"),worst=("max_grad_b","max")).reset_index()
x=np.arange(len(r))
plt.figure(figsize=(6,4))
plt.bar(x-.18,r.mean_max,.36,label="Mean maximum gradient")
plt.bar(x+.18,r.worst,.36,label="Worst gradient")
plt.xticks(x,r.tau.astype(str)); plt.xlabel(r"$\tau_g$")
plt.ylabel(r"$||\nabla_b||$")
plt.title("Multi-seed robustness")
plt.legend(); plt.grid(axis="y",alpha=.25); plt.tight_layout(); plt.show()

display(Markdown(f"""
### Final stability configuration

The forward-changing adaptive-ridge alternatives were rejected because they
either amplified gradients or degraded difficult extrapolation regimes.

Condition-aware backward damping preserved the forward computation and provided
the best stability–fidelity trade-off.

\[
\boxed{{\tau_g={TAU_SELECTED:.2f}}}
\]

This threshold remains provisional until validation inside larger trainable
models.
"""))
```
